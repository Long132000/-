## Задача 1
Написать программу, которая будет выводить на экран числа Фибоначчи, которые меньше переданного числа.

```python
def fibonacci_less_than(n):
    """
    Выводит числа Фибоначчи, которые меньше заданного числа n
    """
    if n <= 0:
        print("Число должно быть положительным")
        return
    
    # Первые два числа Фибоначчи
    a, b = 0, 1
    fibonacci_numbers = []
    
    # Генерируем числа Фибоначчи, пока они меньше n
    while a < n:
        fibonacci_numbers.append(a)
        a, b = b, a + b
    
    # Выводим результат
    print(f"Числа Фибоначчи меньше {n}: {fibonacci_numbers}")
    return fibonacci_numbers

# Примеры использования
fibonacci_less_than(10)
fibonacci_less_than(50)
fibonacci_less_than(100)
```

## Задача 2
Напишите программу, которая на вход принимает список чисел и возвращает только уникальные значения из этого списка.

```python
def get_unique_numbers(numbers):
    """
    Возвращает уникальные значения из списка чисел
    """
    # Преобразуем список в множество для удаления дубликатов, затем обратно в список
    unique_numbers = list(set(numbers))
    return unique_numbers

# Пример использования
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5, 6]
unique = get_unique_numbers(numbers)
print(f"Исходный список: {numbers}")
print(f"Уникальные значения: {unique}")
```

## Задача 3
Напишите программу, которая переводит температуру в градусах Фаренгейта в градусы Цельсия.


```python
def fahrenheit_to_celsius(fahrenheit):
    """
    Переводит температуру из градусов Фаренгейта в градусы Цельсия
    Формула: C = (F - 32) * 5/9
    """
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius

# Пример использования
f_temp = 77
c_temp = fahrenheit_to_celsius(f_temp)
print(f"{f_temp}°F = {c_temp:.2f}°C")
```

## Задача 4
Напишите программу, которая проверяет, является ли строка палиндромом без использования стандартных методов.

```python
def is_palindrome(s):
    """
    Проверяет, является ли строка палиндромом без использования стандартных методов
    """
    # Приводим строку к нижнему регистру и убираем пробелы
    cleaned = ''
    for char in s:
        if char != ' ':
            cleaned += char.lower()
    
    # Проверяем, является ли очищенная строка палиндромом
    left = 0
    right = len(cleaned) - 1
    
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    
    return True

# Примеры использования
test_strings = [
    "А роза упала на лапу Азора",
    "мадам",
    "топот",
    "hello",
    "12321",
    "не палиндром"
]

for s in test_strings:
    result = is_palindrome(s)
    print(f"'{s}' -> {'Палиндром' if result else 'Не палиндром'}")
```
