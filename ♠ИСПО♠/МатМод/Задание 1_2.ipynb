## Задача 1

Напишите программу, которая будет выводить на экран площадь квадрата и его периметр. Сторона квадрата будет равна 25.



```python
# Задаем сторону квадрата
side = 25

# Вычисляем площадь квадрата
area = side * side

# Вычисляем периметр квадрата
perimeter = 4 * side

# Выводим результаты
print(f"Площадь квадрата: {area}")
print(f"Периметр квадрата: {perimeter}")
```

## Задача 2
Напишите функцию, которая будет принимать два числа и считать сумму квадратов этих двух чисел, запишите своё вычисление в переменную squared. (Подсказка: Помните, чтобы записать результат функции в переменную, она должна этот результат внутри себя возвращать! Return). 



```python
def sum_of_squares(a, b):
    """
    Функция вычисляет сумму квадратов двух чисел
    """
    squared = a**2 + b**2
    return squared

# Пример использования функции
result = sum_of_squares(3, 4)
print(f"Сумма квадратов: {result}")  # Выведет: Сумма квадратов: 25
```

## Задача 3
Напишите функцию, которая будет принимать на вход список  [“male“,“male“,“female“,“male“,“male“,“female“,”female“] и возвращать его в обратном порядке. Запишите список в переменную new_list. Подсказка: Вам поможет обратная индексация к спискам ! [::-1]. 



```python
def reverse_gender_list(input_list):
    """
    Функция принимает список и возвращает его в обратном порядке
    """
    new_list = input_list[::-1]
    return new_list

# Исходный список
gender_list = ["male", "male", "female", "male", "male", "female", "female"]

# Использование функции
reversed_list = reverse_gender_list(gender_list)

# Вывод результатов
print(f"Исходный список: {gender_list}")
print(f"Перевернутый список: {reversed_list}")
```

## Задача 4
Напишите функцию, которая будет принимать на вход список (new_list) и идти по нему в цикле. На выходе функция должна вывести на экран: Кол-во мужчин 4, Кол-во женщин 3.



```python
def count_genders(gender_list):
    """
    Функция подсчитывает количество мужчин и женщин в списке
    """
    male_count = 0
    female_count = 0
    
    # Идем по списку в цикле
    for gender in gender_list:
        if gender == "male":
            male_count += 1
        elif gender == "female":
            female_count += 1
    
    # Выводим результат на экран
    print(f"Кол-во мужчин {male_count}, Кол-во женщин {female_count}")

# Использование функции с перевернутым списком из предыдущей задачи
new_list = ["female", "female", "male", "male", "female", "male", "male"]
count_genders(new_list)
```

## Задача 5
Напишите функцию, которая будет принимать строку (Например: “female”), а возвращать словарь где ключи — это символы строки, а значения кол-во символов в строке. 



```python
def count_characters(input_string):
    """
    Функция принимает строку и возвращает словарь,
    где ключи - символы строки, а значения - количество каждого символа
    """
    char_count = {}
    
    # Проходим по каждому символу строки
    for char in input_string:
        if char in char_count:
            # Если символ уже есть в словаре, увеличиваем счетчик
            char_count[char] += 1
        else:
            # Если символа нет в словаре, добавляем его со значением 1
            char_count[char] = 1
    
    return char_count

# Пример использования функции
result = count_characters("female")
print(result)
```

## Задача 6
Воспользуйтесь функцией range в Python и выведите на экран сумму квадратов от 1 до 100 (включительно). 



```python
# Вычисляем сумму квадратов от 1 до 100 включительно
sum_of_squares = 0

# Используем range от 1 до 101 (включительно)
for number in range(1, 101):
    sum_of_squares += number ** 2

# Выводим результат
print(f"Сумма квадратов от 1 до 100: {sum_of_squares}")
```

## Задача 7
Укажите какой метод добавит значение в конец списка. Приведите пример.



```python
#Метод append() добавляет значение в конец списка.

# Создаем исходный список
my_list = [1, 2, 3]
print(f"Исходный список: {my_list}")

# Добавляем значение в конец списка с помощью метода append()
my_list.append(4)
print(f"После добавления 4: {my_list}")

# Добавляем еще одно значение
my_list.append(5)
print(f"После добавления 5: {my_list}")

# Можно добавлять разные типы данных
my_list.append("hello")
print(f"После добавления строки: {my_list}")

# Можно добавлять другие списки (как один элемент)
my_list.append([6, 7])
print(f"После добавления вложенного списка: {my_list}")
```

## Задача 8
Как объединить два списка "a" и "b" в один новый список "c"? Приведите пример.



```python
# Оператор + (наиболее распространенный способ)

# Исходные списки
a = [1, 2, 3]
b = [4, 5, 6]

# Объединяем списки
c = a + b

print(f"Список a: {a}")
print(f"Список b: {b}")
print(f"Объединенный список c: {c}")
```

## Задача 9
Укажите какой метод посчитает количество значений 'a' в списке ['a', 'b', 'c', 'e', 'a', 'g', 'a', 'q']. Приведите пример.



```python
#Метод count() подсчитывает количество указанных значений в списке.

# Исходный список
my_list = ['a', 'b', 'c', 'e', 'a', 'g', 'a', 'q']

# Подсчитываем количество букв 'a' с помощью метода count()
count_a = my_list.count('a')

print(f"Список: {my_list}")
print(f"Количество букв 'a' в списке: {count_a}")
```
